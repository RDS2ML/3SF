## EDA 1 

```python
import pandas as pd
from pprint import pprint
```

### **read data**

```python
wine_df = pd.read_csv('data/wine.csv', sep=',')
wine = wine_df.copy()
```

```python
wine.info()
```

### **data cleaning**

```python
# drop duplicate
mask = wine.duplicated()
print('before drop duplicat', wine.shape[0])
wine.drop_duplicates(inplace=True)
print('after  drop duplicat', wine.shape[0])
# nan/fillna
col_nan = {i:round((wine[i].isna().sum()/wine[i].shape[0])*100, 2) for i in wine.columns if wine[i].isna().sum() > 0}
col_nan = dict(sorted(col_nan.items(), key=lambda x: x[1], reverse=True))
print('before cleaning ')
pprint(col_nan)
# drop feature
wine.drop(labels='region_2', axis=1, inplace=True)
# fillna 'unknown'
wine.loc[wine['region_1'].isna(), 'region_1'] = 'unknown'
wine.loc[wine['designation'].isna(), 'designation'] = 'unknown'
wine.loc[wine['taster_twitter_handle'].isna(), 'taster_twitter_handle'] = '@unknown'
wine.loc[wine['taster_name'].isna(), 'taster_name'] = 'unknown'
# fillna mean
wine.loc[wine['price'].isna(), 'price'] = float(wine['price'].mean())
# fillna mode
wine.loc[wine['country'].isna(), 'country'] = str(wine['country'].mode())
wine.loc[wine['province'].isna(), 'province'] = str(wine['province'].mode())
wine.loc[wine['variety'].isna(), 'variety'] = str(wine['variety'].mode())
# check
check = {i:round((wine[i].isna().sum()/wine[i].shape[0])*100, 2) for i in wine.columns if wine[i].isna().sum() > 0}
check = dict(sorted(check.items(), key=lambda x: x[1], reverse=True))
print('after cleaning ')
pprint(check)
```
