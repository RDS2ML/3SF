```python
import pandas as pd
import statistics as stats
```

```python
wine = pd.read_csv('data/wine_cleared.csv', sep=',', index_col=False)
```

```python
# numeric features describe
wine[['points', 'price']].describe()
```

## *statistics*

```python
# wine price mean with useing statistics module
print('Statistics method wine price mean - ', round(stats.mean(wine['price']), 2))
# wine price mean with python methods
print('Python method wine price mean - ', round(sum(wine['price'])/len(wine['price']), 2))
```
